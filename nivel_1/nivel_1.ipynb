```python
import pandas as pd
import json
arquivo = open("../dados/dados_nivel_1.json")
dados = json.load(arquivo)
taxa_cambio = dados["taxa_cambio_usd_brl"]
df = pd.DataFrame(dados["operacoes"])
print(df)
```

Depois de ler o arquivo JSON, coloquei as operações em um DataFrame do Pandas para conseguir visualizar e trabalhar com os dados em formato de tabela.

Analisando a tabela, encontrei alguns problemas: uma operação duplicada, uma operação sem data e uma operação em USD.

```python
df = df.drop_duplicates()
print(df)
```

Identifiquei que a `OP-0007` estava duplicada. Como os dois registros eram iguais, removi a duplicata para que a mesma operação não fosse contabilizada duas vezes nas análises

Usei drop_duplicates() pra remover a repetição. 
Agora ficaram 19 operações únicas.

```python
df["valor_brl"] = df["valor"].astype(float)
df.loc[df["moeda"] == "USD", "valor_brl"] = df["valor"] * taxa_cambio
print(df)
```

### Erro que aconteceu durante o desenvolvimento

Percebi que a ordem das etapas estava errada. Então passei a criar primeiro a coluna `valor_brl` e depois o `df_com_data`, garantindo que a cópia já tivesse essa coluna.

```python
df_com_data = df[df["data"].notna()].copy()
```

```python
soma_por_cliente_dia = df_com_data.groupby(["cliente_id", "data"])["valor_brl"].sum()
print(soma_por_cliente_dia)
```

```python
suspeitos_fracionamento = soma_por_cliente_dia[soma_por_cliente_dia > 50000]
print(suspeitos_fracionamento)
```

```python
contagem_operacoes = df_com_data.groupby(["cliente_id", "data"])["id"].count()
suspeitos_fracionamento = soma_por_cliente_dia[(soma_por_cliente_dia > 50000) & (contagem_operacoes > 1)]
print(suspeitos_fracionamento)
```

Um cliente é sinalizado se, no mesmo dia, a soma das operações passar de R$ 50.000 e ele tiver mais de uma operação naquele dia.

Percebi que, sem essa segunda condição, a regra também pegava clientes com uma única operação grande. Mas isso não seria fracionamento, e sim outro tipo de suspeita, que será analisado na Regra 2.

Por isso, coloquei essa condição para a regra fazer sentido com a ideia de fracionamento, que é dividir um valor grande em partes menores.

No resultado, o CLI-A-1 foi sinalizado, com 3 operações somando R$ 54.200, e o CLI-A-2 também, com 2 operações somando R$ 52.900.

```python
mediana_por_cliente = df.groupby("cliente_id")["valor_brl"].median()
print(mediana_por_cliente)
```

```python
df["mediana_cliente"] = df["cliente_id"].map(mediana_por_cliente)
df["valor_atipico"] = df["valor_brl"] > (df["mediana_cliente"] * 3)
print(df[["cliente_id", "valor_brl", "mediana_cliente", "valor_atipico"]])
```

## Regra 2: Valor atípico

Calculei a mediana de valor por cliente, usando ela como referência do que seria um comportamento mais normal daquele cliente. Uma operação é marcada como atípica se for maior que 3x essa mediana.

Usei a mediana em vez da média porque ela não é tão afetada por valores muito altos isolados. Por exemplo, a média do CLI-A-4 seria puxada pra cima pela própria operação de R$ 64.800, o que poderia distorcer a comparação.

O resultado foi que só o CLI-A-4 foi sinalizado, na operação de R$ 64.800, que é quase 12x a mediana dele de R$ 5.450.

Essa é a mesma operação que tinha ficado de fora da Regra 1 depois que refinei a regra. Isso faz sentido porque ela não era um caso de fracionamento, já que foi uma operação única e grande.

```python
print("Caso 1: CLI-A-1 - DEVE ser sinalizado (fracionamento real)")
print(df[df["cliente_id"] == "CLI-A-1"][["id", "data", "valor_brl"]])
print("Soma no dia 09/03:", soma_por_cliente_dia["CLI-A-1", "2026-03-09"])

print("")

print("Caso 2: CLI-A-3 - NÃO deve ser sinalizado (após remover duplicata)")
print(df[df["cliente_id"] == "CLI-A-3"][["id", "data", "valor_brl"]])
print("Soma no dia 05/03:", soma_por_cliente_dia["CLI-A-3", "2026-03-05"])
```

Comparei dois casos parecidos pra confirmar que a regra funciona certo.

**CLI-A-1** (dia 09/03): 3 operações somando R$ 54.200. Como passa de R$ 50 mil, foi corretamente sinalizado como fracionamento.
